## 분석
1. soil type과 crop type 두 케이스 모두 temperature, humidity와 target간의 상관관계는 찾지 못함
2. Moisture의 경우 28-28은 눈에 띄게 습도가 높은 지역에서 많이 사용되었으며 17-17-17, 10-26-26, 14-35-14 3가지는 습도 높은 지역에서 거의 사용되지 않음 단, 그중에 14-35-14는 crop type에 따라 습도 높은 지역에서도 사용되는 경우가 있는 것으로 보임 
3. Nitrogen도 유의미한 차이는 없어보임
4. Potassium은 그 차이가 꽤 심함 28028, DAP, Urea는 Potassium이 낮은 곳에서 확연히 많이 쓰이고 나머지는 그 반대
5. Phophorous는 일관된 차이가 보이진 않음

```python
def map_at_3(y_true, y_pred_proba, classes):
    # 상위 3개 예측 클래스의 인덱스를 가져옴
    top_3_preds_idx = np.argsort(-y_pred_proba, axis=1)[:, :3]
    
    # 인덱스를 실제 클래스 라벨로 변환
    top_3_preds_labels = classes[top_3_preds_idx]
    
    # 점수 계산
    score = 0
    for i, true_label in enumerate(y_true):
        pred_labels = top_3_preds_labels[i]
        if true_label in pred_labels:
            # 정답이 예측 리스트에 있으면, 그 위치(rank)에 따라 점수 추가
            rank = np.where(pred_labels == true_label)[0][0] + 1
            score += 1 / rank
            
    return score / len(y_true)
```

```python
from sklearn.pipeline import make_pipeline
from src.features import InteractionFeatureGenerator, TargetEncoder, RatioFeatureGenerator

# Pipeline -> Baseline model

# 타겟 정의  
X = train_df.drop(columns='Fertilizer Name')
y = train_df['Fertilizer Name']

# 전처리기 & 피처 생성기 정의 
interaction_fe = InteractionFeatureGenerator()
ratio_fe = RatioFeatureGenerator()

# target encoding 적용 범주형 컬럼 지정 
target_encoding = TargetEncoder(categorical_features=categorical_data)

# 피처 생성기
# preprocessor = make_column_transformer(
#    (OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_data),
#    remainder='passthrough'
# )

# 모델 정의 
model = LGBMClassifier(random_state=44)

# 파이프라인 생성 
## 전처리기와 모델을 하나의 기계로 묶음 
pipeline = make_pipeline(
    interaction_fe,
    ratio_fe,
    target_encoding,
    # preprocessor, 
    model
)

# 교차 검증 및 평가 
N_SPLITS = 5
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=44)
cv_scores = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    pipeline.fit(X_train, y_train)
    
    # 1. predict 대신 predict_proba를 사용해서 '확률'을 예측
    pred_probas = pipeline.predict_proba(X_val)
    
    # 2. 방금 만든 map_at_3 함수로 점수 계산
    # pipeline['lgbmclassifier'].classes_ 는 모델이 학습한 클래스의 순서를 담고 있음
    score = map_at_3(y_val.values, pred_probas, pipeline['lgbmclassifier'].classes_)
    cv_scores.append(score)
    
    print(f'Fold {fold+1} MAP@3: {score: .5f}')
    
# 결과
mean_cv_score = np.mean(cv_scores)
std_cv_score = np.std(cv_scores)

print(f'Baseline CV Score: {mean_cv_score: .5f} +/- {std_cv_score: .5f}')
```
